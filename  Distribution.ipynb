```python
import pandas as pd
import seaborn as sns
df=sns.load_dataset("tips")
#df["total_bill"].value_counts()
df.sample(30)
```

```python
sns.histplot(data=df,x="total_bill")
```

```python
sns.displot(data=df,kind="hist",x="total_bill")
```

```python
# bins parameter
sns.displot(data=df,kind="hist",x="total_bill",bins=3)
```

```python
# its also possible to visualize  the distribution of a categorical variable using the logic of a histogram 
# discrete bins are automatically set for categorical variable
sns.displot(data=df,kind="hist",x="day")  #how many costumer come in restorent each day 
```

```python
# hue parameter
sns.displot(data=df,kind="hist",x="tip",hue="sex")
```

```python
# element-> step
sns.displot(data=df,kind="hist",x="tip",hue="sex",element="step")
```

```python
titanic=sns.load_dataset("titanic")
titanic
```

```python
sns.displot(data=titanic,kind="hist",x="age",hue="sex",element="step") 
```

```python
sns.displot(data=titanic,kind="hist",x="age",col="sex",element="step")
```

```python
# kde plot(kernal density estimation) ek continuous probability distribution  ko smooth curve ke
# form me show kaeta hai
sns.kdeplot(data=df,x="total_bill")  # maens this graph 10 to 22 ki probability 53% approxmetly hai 
```

```python
sns.displot(kind="kde",data=df,x="total_bill") # this is a figure lavel function
```

```python
# hue parameter
sns.displot(kind="kde",data=df,x="total_bill",hue="sex",fill=True,color="black",linewidth=2)
```

```python
# height lenght increase and aspect  is width( displot me figure ko increase and dicreage nahi kar sakte hai)
sns.displot(kind="kde",data=df,x="total_bill",hue="sex",fill=True,color="black",linewidth=2,height=12,aspect=3)
```

# rugplot

```python
  # rugplot marginal distribution by drawing ticks along the x and y axis
```

```python
sns.kdeplot(data=df,x="total_bill")
sns.rugplot(data=df,x="total_bill")
```

```python
# bivariate histogram
# A bivariate histogram bins the data  within rectangle that tile the plot
# and show the count of observation within each  ractangle with the fill color
sns.histplot(data=df ,x="total_bill",y="tip")  # means 15 doller total bill and  2.2 doller tip sabse jayada hai
```

```python
sns.displot(data=df ,kind="hist",x="total_bill",y="tip") 
```

```python
# bivariate kdeplot
#a bivariate kde plot smoooth the (x,y) observation with a 2D gaussion
sns.kdeplot(data=df,x="total_bill",y="tip") 
```
