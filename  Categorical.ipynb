```python
import seaborn as sns
import plotly.express as px
```

```python
df=sns.load_dataset("tips")
```

```python
df
```

```python
sns.scatterplot(data=df,x="total_bill",y="tip")
```

```python
# stripplot-> work will be categorical data vs numerical data
# axis lavel function 
sns.stripplot(data=df,x="day",y="total_bill")
```

```python
sns.stripplot(data=df,x="day",y="total_bill",jitter=False,hue="sex")
```

```python
# using catplot
# this is a figure flavel function
sns.catplot(data=df,kind="strip",x="day",y="total_bill",hue="sex")
```

```python
sns.catplot(data=df,x="day",kind='strip',y="total_bill",hue='sex',col='time')
# esme col_wrap bhi pass kar sakte hi
```

```python
sns.catplot(data=df,x="day",kind='strip',y="total_bill",col='sex',hue='sex')
```

```python
#swarmplot-> is a axis lavel function
# swarmpot  generally small data set pe work karta hai
# Swarm plot Seaborn ka categorical plot hai. Ye category ke andar har individual data point ko dikhata hai,
#aur points ko thoda side-by-side spread karta hai taaki overlap kam ho.
sns.swarmplot(data=df,x="day",y="total_bill")
```

```python
sns.catplot(kind="swarm",data=df,x="day",y="total_bill",hue='sex',col='time')
```

## Categorical estimate plot

```python
# Barplot
# pointplot
# countplot
```

```python
# barplot -> Bar plot is a graph that represents the average value of a numerical variable for different
# categories using rectangular bars.
sns.barplot(data=df,x='sex',y="total_bill",hue='sex')
```

```python
# when the multiple observation in each categories,it alse use bootstraping a confidence interval the estimate,
# which  is plotted error bars
```

```python
sns.barplot(data=df,x='sex',y="total_bill",hue='smoker') # bydefault mean hai
```

```python
# barplot , ek categorical data ka ek numerical data ke sath ek particuller estimate (mean,min,max,sum,median)
# calculate kar ke deta ahi
sns.barplot(data=df,x='sex',y="total_bill",hue='smoker',estimator="min")
```

```python




sns.catplot(data=df,x='sex',y="total_bill",hue='smoker',estimator="max",kind="bar")
```

```python
# countplot-> A special case for the barplot is when you want  to show the number of obsrvation in each 
# category rather then the computing a statistic for a second numerical variable
sns.countplot(data=df,x="sex",hue="sex")
```

```python
sns.countplot(data=df,x="sex",hue="day")
```

```python
sns.catplot(kind="count",data=df,x="sex",hue="day")
```

```python
# feceting catploy
sns.catplot(kind="box",x='sex',y="total_bill",data=df,hue='sex',row='time',col='smoker',palette="Set2")
```
