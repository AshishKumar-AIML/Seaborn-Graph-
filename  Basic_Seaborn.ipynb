```python
import plotly.express as px  # used for import dataset
```

```python
# scatter plot  -> Axis lavel fuction
sns.scatterplot(data=df,x="total_bill",y="tip",hue="smoker",legend=True )  # legend=True bydefault hai
# hue parameter categori ke according color dega 
```

```python
sns.scatterplot(data=df,x="total_bill",y="tip",hue="sex",style="time") # style= marker shape
```

```python
#  relplot is a figure lavel function
sns.relplot(data=df,x="total_bill", y="tip",kind="scatter",hue="sex",style="time")
```

```python
temp_df=gap[gap["country"]=="India"]
temp_df
```

```python
# lineplot , axis lavel 
sns.lineplot(data=temp_df,x="year",y="lifeExp")
```

```python
# figure lavel function
sns.relplot(data=temp_df,x="year",y="lifeExp",kind="line")
```

```python
# hue-> style
temp_df1=gap[gap["country"].isin(["India","Pakistan","Germany","Brazil","China"])]
temp_df1
```

```python
sns.relplot(data=temp_df1,x="year",y="lifeExp",kind="line",hue="country",style="continent",size="continent")
```

```python
sns.lineplot(data=temp_df1,x="year",y="lifeExp",hue="country")  # axis lavel function
```

```python
# facet plot
sns.relplot(data=df,x="total_bill",y="tip",kind="scatter",hue="sex") # esme male and female ka garph alag alag show ho the 
# i pass fecet(col,row) parameter
```

```python
sns.relplot(data=df,x="total_bill",y="tip",kind="scatter",col="sex") 
```

```python
# fecet plot will work only with figure lavel function-> 
# it will not work with scatter and line plot( because this is a axis lavel function)
sns.relplot(data=df,x="total_bill",y="tip",kind="scatter",col="sex",row="smoker") 
```
