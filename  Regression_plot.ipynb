```python
import seaborn as sns
import pandas as pd
```

```python
df=sns.load_dataset("tips")
```

```python
sns.scatterplot(data=df,x="total_bill",y="tip")
```

```python
# axis lavel plot
sns.regplot(data=df,x="total_bill",y="tip") # hue paramete is not availble
```

```python
sns.lmplot(data=df,x="total_bill",y="tip",hue="sex") # hue parameter is avilble
```

```python
#### a-> Joint Plot
#### b-> Pair plot
```

```python
df1=sns.load_dataset("iris")
df1
```

```python
# pairplot ( pairplot autometically numerical columns ko ditect karta hai
# and pair by pair all columns pe scatter and hist graph plot karega
sns.pairplot(df1)
```

```python
sns.histplot(data=df1,x="petal_width")
```

```python
sns.pairplot(df1,hue="species")
```

```python
sns.kdeplot(data=df1,x="petal_width",hue="species")
```

```python
# pair grid
g=sns.PairGrid(data=df1,hue="species")
g.map(sns.scatterplot)
```

```python
# PairGrid
g=sns.PairGrid(data=df1,hue="species")
# g.map()
g.map(sns.histplot)
```

```python
# map_diag-> map_offdiag
g = sns.PairGrid(data=df1,hue='species')
g.map_diag(sns.boxplot)
g.map_offdiag(sns.scatterplot)
```

```python
g = sns.PairGrid(data=df1,hue='species')
g.map_diag(sns.boxplot)
g.map_offdiag(sns.histplot)
```

```python
# map_diag -> map_upper -> map_lower
g = sns.PairGrid(data=df1,hue='species')
g.map_diag(sns.histplot)
g.map_upper(sns.kdeplot)
g.map_lower(sns.scatterplot)
```

```python
# vars
g = sns.PairGrid(data=df1,hue='species',vars=['sepal_width','petal_width'])
g.map_diag(sns.histplot)
g.map_upper(sns.kdeplot)
g.map_lower(sns.scatterplot)
```
