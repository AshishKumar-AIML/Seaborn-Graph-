# Matrix plot

```python
# Heatmap
# plot ractangular data as a color encoded matrix
import seaborn as sns
import plotly.express as px
```

```python
gap=px.data.gapminder()
gap
```

```python
# pivot function  means row and columns ko rearrange karke summary table banana
temp_df=gap.pivot(index="country",columns="year",values="lifeExp")
temp_df
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(15,15))
sns.heatmap(temp_df)
```

```python
dff=gap[gap["continent"]=="Europe"]
dff
```

```python
temp=dff.pivot(index="country",columns="year",values="lifeExp" )
plt.figure(figsize=(15,15))
sns.heatmap(temp,annot=True,linewidth=1)
```

```python
plt.figure(figsize=(15,15))
sns.heatmap(temp,annot=True,linewidth=1,cmap="viridis")
```
